```python
#lendo a coluna BasePay e nota que ela tem valores como object que precisa ser tratado
#precisamos de valores em float para poder fazer as análises corretas.
df["BasePay"]
```

```python
# valores true é onde são Not provided.
df["BasePay"] == "Not Provided"
```

```python
#agora fazendo uma filtragem condicional
#esses valores fazem com que o pandas trate esses valores como texto, onde devem ser int ou float.
df[df["BasePay"] == "Not Provided"]
```

```python
#pegando os dados contrários da condição anterior
#observe que adicionamos o ~ seguido de parentêses dentro do código.
df[~(df["BasePay"] == "Not Provided")]
```

```python
# agora colocamos o resultado que queremos dentro do nosso DataFrame.
df = df[~(df["BasePay"] == "Not Provided")]
```

```python
# não temos mais os valores Not Provided
df["BasePay"]
```

```python
#Quantas pessoas tem a palavra chefe em seu cargo?
#palavra Chief
# aplicando função em toda coluna "JobTitle"

def chefe(valor):
  if "chief" in valor.lower():#.lower transforma chefe em minúsculo
    return True
  else:
    return False

```

```python
#aplicando a função aplly

df["JobTitle"].apply(chefe)
```

```python
#Seleção condicional
df[df["JobTitle"].apply(chefe)]
```
